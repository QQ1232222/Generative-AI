```python
%matplotlib inline

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**3 - np.sin(x)
```

---
## 1. 描點

今天的函數：
$$f(x) = x^3 - \sin(x)$$

```python
x = np.linspace(-10, 10, 100)   # 在 -10 到 10 之間，取 100 個點
y = x**3 - np.sin(x)

plt.plot(x, y)
plt.show()
```

看起來是一條平滑的曲線。

**但其實不是。** matplotlib 做的事情只有一件：把你給的點，用直線一段一段連起來。

我們把點畫出來看看

(怎麼定義函數，不需要每次重打y=...)
(我想用星星符號標點，顏色想用玫瑰金)

```python
# 只取 5 個點
x5 = np.linspace(-10, 10, 5)
plt.plot(x5, f(x5), marker='*', color='#B76E79') # 把每個點用*標出來，顏色改為玫瑰金
plt.show()
```

原來是折線！

那到底要幾個點才夠？我們一次比較看看。

```python
x5 = np.linspace(-10, 10, 5)
plt.plot(x5, f(x5), marker='*')

x10 = np.linspace(-10, 10, 10)
plt.plot(x10, f(x10), marker='*')

x30 = np.linspace(-10, 10, 30)
plt.plot(x30, f(x30), marker='*')

x100 = np.linspace(-10, 10, 100)
plt.plot(x100, f(x100), marker='*')

plt.show()
```

```python
for n in [5, 10, 30, 100]:
    print(n)
```

```python
for n in [5, 10, 30, 100]:
    xn = np.linspace(-10, 10, n)
    plt.plot(xn, f(xn), marker='o', markersize=(1/n)*30, label=f'{n} 個點')

plt.legend() #顯示圖例
plt.show()
```

## 2. 線條樣式：顏色、粗細、虛線

`plt.plot()` 有許多參數可以調，可以隨意試試看!

(plt.plot() 有哪些參數可以調，我想讓圖美一點?)

```python
x = np.linspace(-10, 10, 300)

# lw: linewidth ；ls: linestyle
plt.plot(x, x**3 - 3*x, color='orange', lw=2.5, ls='-.', label='f(x) = x³ - 3x')
plt.plot(x, 3*x**2 - 3, color='blue', lw=2, ls='--', label='f\'(x) = 3x² - 3')
plt.plot(x, 6*x, color='gray', lw=1.5, ls=':', alpha=0.8, label='f\'\'(x) = 6x') #

plt.legend() #顯示圖例
plt.show()
```

`label=` 加上 `plt.legend()` 就會自動生出圖例 —— 畫多條線時是必備的。

## 3. 標題與座標軸

一張圖如果沒有標題、座標軸標籤，別人會不知道你在畫什麼。

- 標題：plt.title()
- x、y 軸名稱：plt.xlabel()、plt.ylabel()
- 格線：plt.grid()
- 畫出x、y軸：plt.axhline(), plt.axvline()
- x顯示範圍：plt.xlim()

```python
x = np.linspace(-10, 10, 300)
plt.plot(x, f(x), color='#2E86AB', lw=2.5, label='f(x) = x³ - sin(x)')

plt.title('f(x) = x³-sin(x)函數圖形')      # 標題
plt.xlabel('x')                    # x 軸名稱
plt.ylabel('f(x)')                 # y 軸名稱
plt.legend()
plt.grid(True, alpha=0.3)          # 格線
plt.axhline(0, color='black', lw=0.8)   # 畫出 x 軸
plt.axvline(0, color='black', lw=0.8)   # 畫出 y 軸
plt.xlim(-10, 10)                    # x 的顯示範圍
plt.show()
```

### 一次畫好幾張圖：`plt.subplots()`

把第 1 段的「點數比較」做成並排的四宮格。

```python
fig, axes = plt.subplots(2, 2, figsize=(9, 6))
x = np.linspace(-10, 10, 300)
axes[0,0].plot(x, x)
axes[0,1].plot(x, x**2)
axes[1,0].plot(x, x**3)
axes[1,1].plot(x, x**4)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(9, 6))

for ax, n in zip(axes.flatten(), [5, 10, 30, 100]):
    xn = np.linspace(-2, 2, n)
    ax.plot(xn, f(xn), marker='o', markersize=3, color='#2E86AB')
    ax.set_title(f'n = {n}')
    ax.grid(alpha=0.3)

plt.tight_layout()   # 自動排版，避免標題疊在一起
plt.show()
```

> 注意：在 subplot 裡面，`plt.title()` 要改成 `ax.set_title()`，`plt.xlabel()` 要改成 `ax.set_xlabel()`。

## 4. 中文顯示

標題想打中文，會發生什麼事？先試試看 ——

```python
x = np.linspace(-10, 10, 300)
plt.plot(x, f(x), color='#2E86AB')
plt.title('f(x) = x³-sin(x) 函數圖形')
plt.xlabel('x 軸')
plt.show()
```

原因是 Colab 的環境裡**沒有安裝任何中文字型**，matplotlib 找不到字，只好畫空框框。

解法：**下載字型 → 加進 matplotlib → 設成預設**。

```python
#下載字型
!wget -q https://github.com/notofonts/noto-cjk/raw/refs/heads/main/Sans/OTF/TraditionalChinese/NotoSansCJKtc-Medium.otf
```

```python
import matplotlib as mpl
import matplotlib.font_manager as fm
fm.fontManager.addfont('NotoSansCJKtc-Medium.otf') #加入字型
mpl.rcParams['font.family'] = 'Noto Sans CJK TC' #預設字型
```

```python
# 讓負號正常顯示
mpl.rcParams['axes.unicode_minus'] = False
```

成功。設定一次之後，整個 notebook 的圖都能打中文了。
