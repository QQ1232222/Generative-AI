# 第一週作業：以參數方程式繪製蝴蝶曲線

## 數學公式原理
蝴蝶曲線的參數方程式（參數 $t \in [0, 12\pi]$）：

$$r(t) = e^{\cos(t)} - 2\cos(4t) - \sin^5\left(\frac{t}{12}\right)$$

將極座標半徑 $r(t)$ 轉換為直角座標 $(x, y)$：
* **水平座標**：$x(t) = \sin(t) \cdot r(t)$
* **垂直座標**：$y(t) = \cos(t) \cdot r(t)$

---
### 公式結構解析
* **$e^{\cos(t)}$**：決定蝴蝶翅膀上下不對稱的比例（上翅較大、下翅較圓潤）。
* **$-2\cos(4t)$**：產生四個象限的翅膀輪廓與凹凸紋理。
* **$-\sin^5\left(\frac{t}{12}\right)$**：隨週期微調曲線半徑，使 $t$ 在 $0$ 到 $12\pi$ 之間繞行時，描繪出多層次的重疊翅脈效果。

## 讀入重要套件

```python
%matplotlib inline

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 繪製蝴蝶曲線-程式碼

```python
# 1. 設定參數 t 的範圍 (0 到 12π 取 12000 個點)
t = np.linspace(0, 12 * np.pi, 12000)

# 2. 計算蝴蝶曲線的參數方程式
r = np.exp(np.cos(t)) - 2 * np.cos(4 * t) - (np.sin(t / 12) ** 5)
x = np.sin(t) * r
y = np.cos(t) * r

# 3. 設定背景&繪製曲線
fig, ax = plt.subplots(figsize=(8, 8), facecolor='#12121e')
ax.set_facecolor('#12121e')

 # 設定翅膀顏色
ax.plot(x, y, color='#ff66cc', alpha=0.25, linewidth=0.6)
scatter = ax.scatter(x, y, c=t, cmap='rainbow', s=2, alpha=0.85)

# 4. 圖表標題&等比例設定
ax.set_title("Butterfly Curve: $r(t) = e^{\\cos t} - 2\\cos(4t) - \\sin^5(t/12)$",
             color='white', fontsize=14, pad=15)
ax.axis('equal')
ax.grid(True, linestyle='--', alpha=0.2, color='white')
ax.tick_params(colors='lightgray')

plt.tight_layout()
plt.show()
```
